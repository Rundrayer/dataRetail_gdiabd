```python
#importar liberias
import pandas as pd
from pathlib import Path

BASE = Path('/content/Proyecto_Datos_IA_Fernando_Fuentes_IDS211')

carpetas = [
    'datos/geo',
    'docuemntos/geo',
    'evidencias/geo',
    'visualizaciones',
    'notebooks'
]

for c in carpetas:
  (BASE / c).mkdir(parents=True, exist_ok=True)

(BASE/'README.md').touch()

print("Estrcutura creada")

for p in sorted(BASE.rglob('*')):
  nivel= len(p.relative_to(BASE).parts) -1
  print('  '*nivel + p.name + ('/' if p.is_dir() else ''))
```

```python

#DEFINIR RUTA DEL CSV DE SUCURSALES
ruta = BASE / 'datos'/'geo'/'sucursales_dataretail_guia06.csv'
#DEFINIR RUTA DEL CSV DE DEMANDA
ruta_dem = BASE /'datos'/'geo' / 'demanda_dataretail_guia06.csv'

faltantes = [r for r in (ruta, ruta_dem) if not r.exists()]

if faltantes:
  raise FileNotFoundError(f"Faltan: {faltantes}")
else:
  print("Ambos archivos CSV existen y están listos para usar")
```

```python
#LEER CSV DE SUCURSALES
df = pd.read_csv(ruta)

#LEER CSV DE DEMANDA
df_dm = pd.read_csv(ruta_dem)

#IMPRIMIR INFORMACION DE SUCURSALES
print("\n--------------Archivo Sucursales-------------\n")
print(f"Registros: {len(df)}")
print(f"Columnas: {df.columns.tolist()}")
print(df.head())

#IMPRIMIR INFORMACION DE DEMANDA
print("\n--------------Archivo Demandas-------------\n")
print(f"Registros: {len(df_dm)}")
print(f"Columnas: {df_dm.columns.tolist()}")
print(df_dm.head())
```

```python
#CONVERTIR COORDENADAS A NÚMERO

df['latitud'] = pd.to_numeric(df['latitud'], errors='coerce')
df['longitud'] = pd.to_numeric (df[ 'longitud'], errors='coerce')

df_dm['latitud'] = pd.to_numeric (df_dm [ 'latitud'], errors='coerce')
df_dm['longitud'] = pd.to_numeric (df_dm['longitud'], errors='coerce')

print("Coordenadas convertidas a numérico")
print("----Sucursales -----")
print(df[['municipio', 'latitud', 'longitud']])
```

```python
#VALIDACION DE PLAUSIBILIDAD PARA EL SALVADOR

mask_valida = (
    df['latitud'].between(13, 15) &
    df['longitud'].between(-91, -87)
)

print(f"Total de registros : {len(df)}")
print(f"Registros válidos : {mask_valida.sum()}")
print(f"Registros inválidos : {(~mask_valida).sum()}")

if (~mask_valida).sum()>0:
  print("Registros invalidos: ")
  print(df[(~mask_valida)])
else:
  print("Todos los registros son válidos y se cuentran en el rango adecuado")
```

```python
#Revision de nulos de sucursales

print("Conteo de valores nulos de sucursales: ")
print(df.isnull().sum())

#Revision de duplicados en sucursales

print("Conteo de valores duplicados de sucursales: ")
print(df.duplicated().sum())

#Revisar coordenadas duplicadas en sucursales

duplicado_coo = df.duplicated(subset=['latitud', 'longitud']).sum()
print(f"Coordenadas duplicadas en sucursales:{duplicado_coo} ")


#Revision de nulos en demanda

print("Conteo de valores nulos de demanda: ")
print(df_dm.isnull().sum())

#Revision de duplicados en demanda

print("Conteo de valores duplicados  de demanda: ")
print(df_dm.duplicated().sum())

```

```python
#Exportar Version validada

ruta_salida = BASE/ 'datos'/'sucursales_dataretail_guia06_validado.csv'
df.to_csv(ruta_salida, index=False, encoding='utf-8')
print(f"Archivo validado exportado a {ruta_salida}")

#DESCARGAR
from google.colab import files
files.download(str(ruta_salida))
```
